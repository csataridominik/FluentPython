```python
# nested unpacking:

metro_areas = [
('Tokyo', 'JP', 36.933, (35.689722, 139.691667)),  
('Delhi NCR', 'IN', 21.935, (28.613889, 77.208889)),
('Mexico City', 'MX', 20.142, (19.433333, -99.133333)),
('New York-Newark', 'US', 20.104, (40.808611, -74.020386)),
('São Paulo', 'BR', 19.649, (-23.547778, -46.635833)),
]

print(f'{"":15} | {"latitude":>9} | {"longitude":>9}')
for name, _, _, (lat, lon) in metro_areas:  
    if lon <= 0:  
        print(f'{name:15} | {lat:9.4f} | {lon:9.4f}')



# for lists unpackign can be used, rarely good use case.
```

```python
# In Python, parentheses () are also used for grouping mathematical expressions.
# Without the comma, Python thinks you are just putting a number in parentheses for math.
x = (6)    # Python sees math grouping: x is an integer (6)
y = (6,)   # Python sees the trailing comma: y is a tuple with one element (6,)

print(x,y)
```

```python
# Pattern matching with sequences:
class InvalidCommand(Exception):
    """Error handeling dummy... """
    pass

def handle_command(self, message):
    match message:  
        case ['BEEPER', frequency, times]:  
            self.beep(times, frequency)
        case ['NECK', angle]:  
            self.rotate_neck(angle)
        case ['LED', ident, intensity]:  
            self.leds[ident].set_brightness(ident, intensity)
        case ['LED', ident, red, green, blue]:  
            self.leds[ident].set_color(ident, red, green, blue)
        case _:  # default case, match anything that does not match above
            raise InvalidCommand(message)



# Not jsut a switch/case like in C, it has Pattern Matching Destructuring.


# match (subject):
#    case [pattern] and a optional guard if keyword:


```
